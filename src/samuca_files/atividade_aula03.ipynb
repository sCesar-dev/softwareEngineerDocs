Os histogramas estão apresentando uma distorção à direita(cauda à direita) porque os primeiros dados do x são mais frequentes(execução mais rápida).Creio que o prox passo seria pegar o valor com maior frequencia de contagens, e extraindo ele teriamos o tempo médio padrão, com o desvio padrão feito.

```python
from time import time
from matplotlib import pyplot as plt
import numpy as np

lista_n = [1000, 10000, 100000]
l = []
R = 100

for n in lista_n:
    tic = time()
    ll = []
    for i in range(R):
        tic = time()
        for i in range(n):
            x = 1
        tac = time()
        ll.append(tac - tic)

    l.append(ll)
    #cria um histograma com ciclos dos 3 valores na lista_n
    plt.hist(ll,25)
    #elevo os valores de x a -4 so pra facilitar a visualizacao
    plt.ticklabel_format(axis='x', style='sci', scilimits=(0, 0), useMathText=True)
    plt.show()
    
    ll_array = np.array(ll)
    l_corte = np.percentile(ll_array, 90)
    ll_filtrado = ll_array[ll_array < l_corte]
    mean = np.mean(ll_filtrado)
    standard_deviation = np.std(ll_filtrado)
    coef_var = standard_deviation / mean

    print(f"mean value: {mean}\nstandard_deviation: {standard_deviation}\ncoef_var: {coef_var}")
    plt.hist(ll_filtrado, bins=15, edgecolor='black')
    plt.title(f"n = {n} | CV = {coef_var}")
    plt.xlabel("Time(s)")
    plt.ylabel("Frequence")
    plt.ticklabel_format(axis='x', style='sci', scilimits=(0, 0), useMathText=True)
    plt.show()
```
