```python
from time import time
from matplotlib import pyplot as plt

lista_n = [1000, 10000, 100000]
l = []
R = 100

for n in lista_n:
    tic = time()
    ll = []
    for i in range(R):
        tic = time()
        for i in range(n):
            x = 1
        tac = time()
        ll.append(tac - tic)
    
    l.append(ll)
    #cria um histograma com ciclos dos 3 valores na lista_n
    plt.hist(ll,50)
    #elevo os valores de x a -4 so pra facilitar a visualizacao
    plt.ticklabel_format(axis='x', style='sci', scilimits=(0, -4))
    plt.show()
```

Os histogramas estão apresentando uma distorção à direita(cauda à direita), mas é normal porque os primeiros dados do x são mais frequentes(execução mais rápida).Creio que o prox passo seria pegar o valor com maior frequencia de contagens, e extraindo ele teriamos o tempo médio padrão, com o desvio padrão feito.
