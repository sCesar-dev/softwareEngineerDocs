```python
import matplotlib.pyplot as plt
import cv2

img = cv2.imread('../images/64152.jpg')
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
plt.figure(figsize=(10, 6))

plt.imshow(img)
plt.show()

```

```python
k = 2
n, m, c = img.shape
print(f'linhas: {n}\ncolunas: {m}\ncores: {c}')
```

```python
faixa = []
plano = []

for cor in range(c):
    plano.append(img[:,:,cor])
    plt.imshow(plano[-1], cmap='gray')
    plt.show()
```

```python
faixa = []
plano = []

for cor in range(c):
    plano = img[:, :, cor]

    fig = plt.figure(
        figsize=(3840/300, 2160/300),
        dpi=300,
        frameon=False
    )

    ax = fig.add_axes([0, 0, 1, 1])
    ax.imshow(plano, cmap='gray')
    ax.axis('off')

    fig.savefig(
        f'supermanLogo_{cor}.png',
        dpi=300,
        bbox_inches=None,
        pad_inches=0
    )

    plt.close(fig)
```
