```python
import json
import numpy as np

with open("samuca_files/times.json", "r") as f:
    dados = json.load(f)

lista_n = [int(n) for n in dados.keys()]
l = list(dados.values())

for n in sorted(dados, key=int):
    tempos = np.array(dados[n])
    minimo = tempos.min()
    media = tempos.mean()
    maximo = tempos.max()
    print(f"N = {n}")
    print(f"  Mínimo : {minimo:.6e}")
    print(f"  Média  : {media:.6e}")
    print(f"  Máximo : {maximo:.6e}")
    print("-" * 40)
```

```python
import json
import numpy as np
from matplotlib import pyplot as plt
from pathlib import Path

arquivos_ciclos = [
    "samuca_files/times1.json",
    "samuca_files/times2.json",
    "samuca_files/times3.json",
]
dados_p_ciclo = []
for caminho in arquivos_ciclos:
    with open(caminho, "r") as f:
        dados_p_ciclo.append(json.load(f))

n_ciclos = len(dados_p_ciclo)
lista_n = sorted(int(n) for n in dados_p_ciclo[0].keys())

means_por_ciclo = {c: [] for c in range(n_ciclos)}
stds_por_ciclo = {c: [] for c in range(n_ciclos)}
coef_vars_por_ciclo = {c: [] for c in range(n_ciclos)}

for c in range(len(dados_p_ciclo)):
    for n in lista_n:
        array_n = np.array(dados_p_ciclo[c][str(n)])
        mean = np.mean(array_n)
        std_deviation = np.std(array_n)
        coef_var = std_deviation / mean

        means_por_ciclo[c].append(mean)
        stds_por_ciclo[c].append(std_deviation)
        coef_vars_por_ciclo[c].append(coef_var)
    
plt.figure(figsize=(8, 4))
for c in range(n_ciclos):
    plt.plot(lista_n, coef_vars_por_ciclo[c], marker="o", markersize=5, linestyle="-", label=f"Ciclo {c+1}")
plt.title(f"Grafico do Coef de Var x Tamanho da Entrada (N)")
plt.xlabel("Tamanho da Entrada")
plt.ylabel("Coef da variacao")
plt.grid(True, linestyle="--", alpha=0.5)
plt.legend()
plt.show()

plt.savefig('grafico_cvXn.png', dpi=300, bbox_inches='tight')
```

```python
plt.figure(figsize=(8, 3))
for c in range(n_ciclos):
    plt.plot(lista_n, stds_por_ciclo[c], marker="o", markersize=5, linestyle="-",  label=f"Ciclo {c+1}")
plt.title(f"Gráfico de Desvio Padrão x Tamanho da Entrada (N)")
plt.xlabel("Tamanho da Entrada")
plt.ylabel("Desvio Padrao")
plt.grid(True, linestyle="--", alpha=0.5)
plt.legend()
plt.show()

plt.savefig('grafico_std-devXn.png', dpi=300, bbox_inches='tight')
```

```python
plt.figure(figsize=(8, 3))
for c in range(n_ciclos):
    plt.plot(lista_n, means_por_ciclo[c], marker="o", markersize=5, linestyle="-",  label=f"Ciclo {c+1}")
plt.title(f"Gráfico de Media Aritmetica x Tamanho da Entrada (N)")
plt.xlabel("Tamanho da Entrada")
plt.ylabel("Media Aritmetica")
plt.grid(True, linestyle="--", alpha=0.5)
plt.legend()
plt.show

plt.savefig('grafico_meanXn.png', dpi=300, bbox_inches='tight')
```

```python
import json
import numpy as np
from matplotlib import pyplot as plt
from pathlib import Path

with open("samuca_files/times.json", "r") as f:
    dados = json.load(f)

lista_n = [int(n) for n in dados.keys()]
l = list(dados.values())
coef_vars = []
means = []
stds_deviations = []

plt.figure(figsize=(10, 5))
for n in lista_n:
    ll = dados[str(n)]
    array = np.array(ll)
    mean = np.mean(array)
    means.append(mean)
    
    x = np.full(len(array), n)
    plt.scatter(x, array, alpha=0.6, edgecolors='black', linewidths=0.3, s=40, label=f"Amostras n={n}")
    
plt.scatter(lista_n, means, facecolors='none', edgecolors='red', marker="o", s=25, linewidths=1.5, zorder=4)

plt.plot(lista_n, means, marker="o", markersize=3, linestyle="-")

plt.title(f"Gráfico de Dispersão dos Tempos de Execução x Tamanho da Entrada (N)")
plt.xlabel("Tamanho da Entrada(n)")
plt.ylabel("Media Aritmetica(s)")
plt.xscale("log")
plt.grid(True, which="both", linestyle="--", alpha=0.5)
plt.legend()
plt.tight_layout()
plt.show

plt.savefig('graficoDispersao_meanPexecXn.png', dpi=300, bbox_inches='tight')
```
