Primeiro é executado o codigo execution_time com tasknet

tasknet -c 13 python execution_time.py

Com isso é gerado um .json com os dados de tempo. Após isso eles são extraidos e inseridos para geração dos graficos.

```python
# Lista os arquivos da pasta atual
import os
print("Arquivos aqui:", os.listdir('.'))

print("Arquivos nivel abaixo:", os.listdir('samuca_files'))
# Lista os arquivos da pasta de cima (um nível acima)
print("Arquivos um nível acima:", os.listdir('..'))
```

```python
import json
import numpy as np
from matplotlib import pyplot as plt
from pathlib import Path

with open("samuca_files/times.json", "r") as f:
    dados = json.load(f)

lista_n = [int(n) for n in dados.keys()]
l = list(dados.values())
coef_vars = []
means = []
stds_deviations = []

for n in lista_n:
    ll = dados[str(n)]
    
    bfr_array = np.array(ll)
    
    mean = np.mean(bfr_array)
    means.append(mean)
    std_deviation = np.std(bfr_array)
    stds_deviations.append(std_deviation)
    coef_var = std_deviation / mean
    coef_vars.append(coef_var)
    
plt.figure(figsize=(8, 3))
plt.plot(lista_n, coef_vars, marker="o", markersize=3, linestyle="-")
plt.title(f"Grafico do Coef de Var x Tamanho da Entrada (N)")
plt.xlabel("Tamanho da Entrada")
plt.ylabel("Coef da variacao")
plt.grid(True, linestyle="--", alpha=0.5)
plt.show

plt.savefig('grafico_cvXn.png', dpi=300, bbox_inches='tight')
```

```python
plt.figure(figsize=(8, 3))
plt.plot(lista_n, stds_deviations, marker="o", markersize=3, linestyle="-")
plt.title(f"Gráfico de Desvio Padrão x Tamanho da Entrada (N)")
plt.xlabel("Tamanho da Entrada")
plt.ylabel("Desvio Padrao")
plt.grid(True, linestyle="--", alpha=0.5)
plt.show

plt.savefig('grafico_std-devXn.png', dpi=300, bbox_inches='tight')
```

```python
plt.figure(figsize=(8, 3))
plt.plot(lista_n, means, marker="o", markersize=3, linestyle="-")
plt.title(f"Gráfico de Media Aritmetica x Tamanho da Entrada (N)")
plt.xlabel("Tamanho da Entrada")
plt.ylabel("Media Aritmetica")
plt.grid(True, linestyle="--", alpha=0.5)
plt.show

plt.savefig('grafico_meanXn.png', dpi=300, bbox_inches='tight')
```

```python
plt.figure(figsize=(10, 5))
for n in lista_n:
    ll = dados[str(n)]
    array = np.array(ll)
    
    x = np.full(len(array), n)
    plt.scatter(x, array, alpha=0.35, edgecolors='none', s=40, label=f"Amostras n={n}")
    
plt.scatter(lista_n, means, color="red", marker="o", s=100, zorder=2, label="Media")
plt.plot(lista_n, means, marker="o", markersize=3, linestyle="-")

plt.title(f"Gráfico de Dispersão dos Tempos de Execução x Tamanho da Entrada (N)")
plt.xlabel("Tamanho da Entrada(n)")
plt.ylabel("Media Aritmetica(s)")
plt.xscale("log")
plt.grid(True, which="both", linestyle="--", alpha=0.5)
plt.legend()
plt.tight_layout()
plt.show

plt.savefig('graficoDispersao_meanPexecXn.png', dpi=300, bbox_inches='tight')
```

```python
for n in range(len(means)):
    print(f"Media em {lista_n[n]}: {means[n]:.6e}\nCV: {coef_vars[n]:.6e}")
```
